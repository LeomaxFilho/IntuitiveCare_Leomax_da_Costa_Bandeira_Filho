# Medias

## Operadora

```python
df_operadora = df.groupby('CNPJ').agg({
    'ValorDespesas': ['sum', 'mean', 'count', 'std'],
    'RazaoSocial': 'first'
})

df_operadora.head()
```

```python
df_uf_trimestre = df.groupby(['UF', 'Trimestre'])['ValorDespesas'].mean().reset_index()
df_uf_trimestre.columns = ['UF', 'Trimestre', 'MediaDespesas']
df_uf_trimestre = df_uf_trimestre.sort_values(by='MediaDespesas')
df_uf_trimestre.head()
```
